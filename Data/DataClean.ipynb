```python
import pandas as pd
import re
from pyarabic import araby
import os
```

### Arabic Normalization Function

```python
def NormalizeArabic(text):
    text = text.strip()
    text = re.sub("ى", "ي", text)
    text = re.sub("ؤ", "ء", text)
    text = re.sub("ئ", "ء", text)

    text = re.sub("[إأٱآا]", "ا", text)

    text = re.sub("ة", "ه", text)

    text = re.sub(r"(.)\1+", r"\1\1", text)

    text = araby.strip_tashkeel(text)
    text = araby.strip_diacritics(text)

    text = re.sub('\s+', ' ', text)
    return text.strip()
```

### Remove Punctuation

```python
def RemovePunctuation(text):
     text = re.sub('[%s]' % re.escape("""!"#$%&'()*+,،-./:;<=>؟?@[\]^_`{|}~"""), ' ', text)
     text = re.sub('\s+', ' ', text)

     return text.strip()
```

### Cleaning Function

```python
def CleanText(text):
    text = NormalizeArabic(text)
    text = RemovePunctuation(text)
    return text
```

```python
for file_name in os.listdir(folder_path):
    if file_name.endswith(".csv"):
        
        file_path = os.path.join(folder_path, file_name)
        print("Processing:", file_name)

        data = pd.read_csv(file_path)

        if "text" not in data.columns:
            print(f"Skipping {file_name} (no 'text' column)")
            continue

        data["text"] = data["text"].astype(str).apply(CleanText)

        new_file_path = os.path.join(folder_path, "cleaned_" + file_name)
        data.to_csv(new_file_path, index=False)

        print(f"{file_name} cleaned \n")
```
